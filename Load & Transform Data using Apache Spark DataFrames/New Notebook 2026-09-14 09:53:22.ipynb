Load and Transform Data using Apache Spark DataFrames

```python
%sql
CREATE VOLUME IF NOT EXISTS dbacademy.get_started_de.myfiles;
```

```python
catalog="dbacademy"
schema="get_started_de"
file_name="rows.csv"
download_url = "https://health.data.ny.gov/api/views/jxy9-yhdk/rows.csv"
file_name = "rows.csv"
table_name = "business_data"
path_volume = "/Volumes/" + catalog + "/" + schema + "/" + 'myfiles'
path_table = catalog + "." + schema
print(path_table) # Show the complete path
print(path_volume) # Show the complete path
```

```python
dbutils.fs.cp(f"{download_url}", f"{path_volume}/{file_name}")
```

```python
df_csv = spark.read.csv(f"{path_volume}/{file_name}", header=True, inferSchema=True, sep=",")
display(df_csv)
```

```python
df_csv = df_csv.withColumnRenamed("First Name", "first_name")
df_csv = df_csv.withColumnRenamed("Last Name", "last_name")
df_csv = df_csv.withColumnRenamed("County", "county")
df_csv = df_csv.withColumnRenamed("Sex", "sex")
df_csv = df_csv.withColumnRenamed("Count", "count")
df_csv.printSchema()
```

```python
df_csv.write.mode("overwrite").saveAsTable(f"{path_table}.{table_name}")
```
